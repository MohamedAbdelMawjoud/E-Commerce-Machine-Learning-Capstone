```python
# Relationship / Key Audit

# Create references to datasets
orders = data['olist_orders_dataset']
customers = data['olist_customers_dataset']
order_items = data['olist_order_items_dataset']
payments = data['olist_order_payments_dataset']
reviews = data['olist_order_reviews_dataset']
products = data['olist_products_dataset']
sellers = data['olist_sellers_dataset']


def key_audit(df, column, table_name):
    print("=" * 70)
    print(f"{table_name}.{column}")
    print("=" * 70)
    print(f"Rows: {len(df):,}")
    print(f"Nulls: {df[column].isna().sum():,}")
    print(f"Unique values: {df[column].nunique():,}")
    print(f"Duplicated values: {df[column].duplicated().sum():,}")
    print()


# Primary / Main Keys

key_audit(orders, 'order_id', 'orders')
key_audit(customers, 'customer_id', 'customers')
key_audit(products, 'product_id', 'products')
key_audit(sellers, 'seller_id', 'sellers')

key_audit(reviews, 'review_id', 'reviews')



# Foreign Keys / Relationship Columns

key_audit(orders, 'customer_id', 'orders')

key_audit(order_items, 'order_id', 'order_items')
key_audit(order_items, 'product_id', 'order_items')
key_audit(order_items, 'seller_id', 'order_items')

key_audit(payments, 'order_id', 'payments')

key_audit(reviews, 'order_id', 'reviews')
```

```python
# Preprocessing - Step 1
# Create Modeling Base

orders_model = orders.copy()

print("Original orders:", orders_model.shape)

display(
    orders_model['order_status']
    .value_counts()
    .to_frame('count')
)
```

```python
delivered_orders = orders_model[
    orders_model['order_status'] == 'delivered'
].copy()

print("Delivered orders:", delivered_orders.shape)
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    delivered_orders[col] = pd.to_datetime(
        delivered_orders[col],
        errors='coerce'
    )
```

```python
delivered_orders[date_columns].dtypes
```

```python
delivered_orders['delivery_delay_days'] = (
    delivered_orders['order_delivered_customer_date']
    - delivered_orders['order_estimated_delivery_date']
).dt.total_seconds() / (24 * 60 * 60)
```

```python
delivered_orders['is_late'] = (
    delivered_orders['delivery_delay_days'] > 0
).astype(int)
```

```python
print("Target distribution:")
display(
    delivered_orders['is_late']
    .value_counts()
    .sort_index()
    .to_frame('count')
)

print("\nTarget percentage:")
display(
    delivered_orders['is_late']
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .to_frame('percentage')
)
```

```python
items_model = order_items.copy()
```

```python
items_agg = (
    items_model
    .groupby('order_id')
    .agg(
        total_items=('order_item_id', 'count'),
        total_price=('price', 'sum'),
        total_freight=('freight_value', 'sum'),
        avg_item_price=('price', 'mean'),
        avg_freight=('freight_value', 'mean'),
        unique_products=('product_id', 'nunique'),
        unique_sellers=('seller_id', 'nunique')
    )
    .reset_index()
)
```

```python
print(items_agg.shape)
display(items_agg.head())
```

```python
payments_agg = (
    payments
    .groupby('order_id')
    .agg(
        payment_count=('payment_sequential', 'count'),
        total_payment_value=('payment_value', 'sum'),
        max_installments=('payment_installments', 'max'),
        avg_installments=('payment_installments', 'mean')
    )
    .reset_index()
)
```

```python
print(payments_agg.shape)
display(payments_agg.head())
```

```python
customers_model = customers[
    [
        'customer_id',
        'customer_zip_code_prefix',
        'customer_city',
        'customer_state'
    ]
].copy()
```

```python
# Product-level features → Order-level

products_model = products.copy()

product_item_data = order_items[
    ['order_id', 'product_id']
].merge(
    products_model,
    on='product_id',
    how='left',
    validate='many_to_one'
)

product_agg = (
    product_item_data
    .groupby('order_id')
    .agg(
        unique_categories=('product_category_name', 'nunique'),
        avg_product_weight_g=('product_weight_g', 'mean'),
        avg_product_photos_qty=('product_photos_qty', 'mean'),
        avg_product_name_length=('product_name_lenght', 'mean'),
        avg_product_description_length=('product_description_lenght', 'mean'),
        avg_product_length_cm=('product_length_cm', 'mean'),
        avg_product_height_cm=('product_height_cm', 'mean'),
        avg_product_width_cm=('product_width_cm', 'mean')
    )
    .reset_index()
)

print("Product aggregation shape:", product_agg.shape)

display(product_agg.head())
```

```python
# Seller-level features → Order-level

seller_item_data = order_items[
    ['order_id', 'seller_id']
].merge(
    sellers[
        ['seller_id', 'seller_zip_code_prefix',
         'seller_city', 'seller_state']
    ],
    on='seller_id',
    how='left',
    validate='many_to_one'
)

seller_agg = (
    seller_item_data
    .groupby('order_id')
    .agg(
        unique_sellers=('seller_id', 'nunique'),
        unique_seller_states=('seller_state', 'nunique')
    )
    .reset_index()
)

print("Seller aggregation shape:", seller_agg.shape)

display(seller_agg.head())
```

```python
orders_model = delivered_orders[
    [
        'order_id',
        'customer_id',
        'order_status',
        'order_purchase_timestamp',
        'order_approved_at',
        'order_delivered_carrier_date',
        'order_estimated_delivery_date',
        'delivery_delay_days',
        'is_late'
    ]
].copy()
```

```python
master_df = orders_model.copy()

print("Initial master shape:", master_df.shape)
```

```python
master_df = master_df.merge(
    customers_model,
    on='customer_id',
    how='left',
    validate='one_to_one'
)

print("After customers:", master_df.shape)
```

```python
master_df = master_df.merge(
    items_agg,
    on='order_id',
    how='left',
    validate='one_to_one'
)

print("After order items:", master_df.shape)
```

```python
master_df = master_df.merge(
    payments_agg,
    on='order_id',
    how='left',
    validate='one_to_one'
)

print("After payments:", master_df.shape)
```

```python
# ============================================================
# MODEL 2 — DELIVERY DELAY PREDICTION
# ============================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ============================================================
# 1. FEATURES + TARGET
# ============================================================

target = 'is_late'

features = [
    'estimated_delivery_days',
    'customer_state',
    'customer_city',
    'customer_zip_code_prefix',
    'total_items',
    'total_price',
    'total_freight',
    'avg_item_price',
    'avg_freight',
    'unique_products',
    'unique_sellers',
    'unique_seller_states',
    'payment_count',
    'max_installments',
    'avg_installments',
    'purchase_hour',
    'purchase_month',
    'purchase_weekday',
    'is_weekend',
    'approval_delay_hours'
]

X = master_df[features].copy()
y = master_df[target].copy()

# ============================================================
# 2. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)

# ============================================================
# 3. NUMERICAL / CATEGORICAL FEATURES
# ============================================================

categorical_features = [
    'customer_state',
    'customer_city'
]

numeric_features = [
    col for col in features
    if col not in categorical_features
]

# ============================================================
# 4. ENCODING + PREPROCESSING
# ============================================================

numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=True
    ))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# ============================================================
# 5. MODELS
# ============================================================

models = {
    'Logistic Regression':
        LogisticRegression(
            max_iter=1000
        ),

    'Logistic Regression Balanced':
        LogisticRegression(
            max_iter=1000,
            class_weight='balanced'
        ),

    'Decision Tree':
        DecisionTreeClassifier(
            random_state=42
        ),

    'Decision Tree Balanced':
        DecisionTreeClassifier(
            random_state=42,
            class_weight='balanced'
        ),

    'Random Forest':
        RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ),

    'Random Forest Balanced':
        RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            n_jobs=-1,
            class_weight='balanced'
        ),

    'KNN':
        KNeighborsClassifier(
            n_neighbors=5
        )
}

# ============================================================
# 6. TRAIN + EVALUATE
# ============================================================

results = []
trained_models = {}

for name, model in models.items():

    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)
    y_prob = pipeline.predict_proba(X_test)[:, 1]

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_prob)
    })

    trained_models[name] = pipeline

# ============================================================
# 7. RESULTS
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    'F1',
    ascending=False
).reset_index(drop=True)

display(results_df)

# ============================================================
# 8. FEATURE IMPORTANCE 
# ============================================================
import matplotlib.pyplot as plt

best_tree_pipeline = trained_models['Random Forest']  # أو 'Random Forest Balanced'

#  OneHotEncoding
feature_names = best_tree_pipeline.named_steps['preprocessor'].get_feature_names_out()

importances = best_tree_pipeline.named_steps['model'].feature_importances_

importance_df = pd.DataFrame({
    'feature': feature_names,
    'importance': importances
}).sort_values('importance', ascending=False).head(15)

plt.figure(figsize=(9, 6))
plt.barh(importance_df['feature'], importance_df['importance'])
plt.gca().invert_yaxis()
plt.title('Top 15 Feature Importances — Random Forest (Delivery Delay)')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()

display(importance_df)
```

```python
# TUNING — MODEL 2: Logistic Regression Balanced

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

pipeline_logreg = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

param_grid_logreg = {
    'model__C': [0.01, 0.1, 1],          #regularization
    'model__penalty': ['l2'],
    'model__solver': ['lbfgs', 'liblinear']
}

cv_strat = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid_logreg = GridSearchCV(
    pipeline_logreg,
    param_grid_logreg,
    cv=cv_strat,
    scoring='f1_macro',
    n_jobs=-1,
    verbose=1
)

grid_logreg.fit(X_train, y_train)

print(f'Best params: {grid_logreg.best_params_}')
print(f'Best F1-macro (CV): {grid_logreg.best_score_:.4f}')

best_logreg = grid_logreg.best_estimator_

# تقييم على test set
y_pred_tuned = best_logreg.predict(X_test)
y_prob_tuned = best_logreg.predict_proba(X_test)[:, 1]

print(f'Test Precision: {precision_score(y_test, y_pred_tuned):.4f}')
print(f'Test Recall:    {recall_score(y_test, y_pred_tuned):.4f}')
print(f'Test F1:        {f1_score(y_test, y_pred_tuned):.4f}')
print(f'Test ROC-AUC:   {roc_auc_score(y_test, y_prob_tuned):.4f}')
```
